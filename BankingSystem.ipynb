```python
import random
from datetime import datetime


# ============================================================
# BANKING SYSTEM
# ============================================================

# Dictionary to store all bank accounts
# Format:
# accounts = {
#     account_number: {
#         "name": "John",
#         "phone": "9876543210",
#         "pin": "1234",
#         "balance": 5000.0,
#         "transactions": []
#     }
# }

accounts = {}


# ============================================================
# UTILITY FUNCTIONS
# ============================================================

def generate_account_number():
    """Generate a unique 6-digit account number."""

    while True:
        account_number = str(random.randint(100000, 999999))

        if account_number not in accounts:
            return account_number


def get_current_time():
    """Return the current date and time."""

    return datetime.now().strftime("%d-%m-%Y %H:%M:%S")


def add_transaction(account_number, transaction_type, amount, details=""):
    """Add a transaction to an account's transaction history."""

    transaction = {
        "type": transaction_type,
        "amount": amount,
        "date": get_current_time(),
        "details": details
    }

    accounts[account_number]["transactions"].append(transaction)


def get_valid_amount(message):
    """Get a valid positive amount from the user."""

    while True:
        try:
            amount = float(input(message))

            if amount <= 0:
                print("Amount must be greater than 0.")
            else:
                return amount

        except ValueError:
            print("Please enter a valid number.")


# ============================================================
# CREATE ACCOUNT
# ============================================================

def create_account():
    """Create a new bank account."""

    print("\n" + "=" * 50)
    print("              CREATE BANK ACCOUNT")
    print("=" * 50)

    # Get name
    while True:
        name = input("Enter your name: ").strip()

        if name == "":
            print("Name cannot be empty.")
        elif not all(character.isalpha() or character.isspace() for character in name):
            print("Name should contain only letters and spaces.")
        else:
            break

    # Get phone number
    while True:
        phone = input("Enter your phone number: ").strip()

        if len(phone) == 10 and phone.isdigit():
            break
        else:
            print("Please enter a valid 10-digit phone number.")

    # Create PIN
    while True:
        pin = input("Create a 4-digit PIN: ").strip()

        if len(pin) == 4 and pin.isdigit():
            confirm_pin = input("Confirm your PIN: ").strip()

            if pin == confirm_pin:
                break
            else:
                print("PINs do not match.")
        else:
            print("PIN must contain exactly 4 digits.")

    # Generate account number
    account_number = generate_account_number()

    # Create account
    accounts[account_number] = {
        "name": name,
        "phone": phone,
        "pin": pin,
        "balance": 0.0,
        "transactions": []
    }

    # Add account creation transaction
    add_transaction(
        account_number,
        "ACCOUNT CREATED",
        0,
        "Bank account created successfully"
    )

    print("\n" + "=" * 50)
    print("       ACCOUNT CREATED SUCCESSFULLY!")
    print("=" * 50)
    print(f"Account Holder : {name}")
    print(f"Account Number : {account_number}")
    print(f"Initial Balance: ₹0.00")
    print("=" * 50)
    print("Please remember your account number and PIN.")


# ============================================================
# LOGIN
# ============================================================

def login():
    """Login to an existing bank account."""

    print("\n" + "=" * 50)
    print("                    LOGIN")
    print("=" * 50)

    if not accounts:
        print("No accounts exist yet.")
        print("Please create an account first.")
        return

    account_number = input("Enter account number: ").strip()

    if account_number not in accounts:
        print("Account not found.")
        return

    pin = input("Enter PIN: ").strip()

    if accounts[account_number]["pin"] != pin:
        print("Incorrect PIN.")
        return

    print("\nLogin successful!")
    print(f"Welcome, {accounts[account_number]['name']}!")

    account_menu(account_number)


# ============================================================
# CHECK BALANCE
# ============================================================

def check_balance(account_number):
    """Display current account balance."""

    balance = accounts[account_number]["balance"]

    print("\n" + "=" * 50)
    print("                 ACCOUNT BALANCE")
    print("=" * 50)
    print(f"Account Number : {account_number}")
    print(f"Account Holder : {accounts[account_number]['name']}")
    print(f"Current Balance: ₹{balance:.2f}")
    print("=" * 50)


# ============================================================
# DEPOSIT
# ============================================================

def deposit_money(account_number):
    """Deposit money into the account."""

    print("\n" + "=" * 50)
    print("                   DEPOSIT")
    print("=" * 50)

    amount = get_valid_amount("Enter amount to deposit: ₹")

    accounts[account_number]["balance"] += amount

    add_transaction(
        account_number,
        "DEPOSIT",
        amount,
        "Money deposited into account"
    )

    print("\nDeposit successful!")
    print(f"Deposited Amount : ₹{amount:.2f}")
    print(f"New Balance      : ₹{accounts[account_number]['balance']:.2f}")


# ============================================================
# WITHDRAW
# ============================================================

def withdraw_money(account_number):
    """Withdraw money from the account."""

    print("\n" + "=" * 50)
    print("                  WITHDRAW")
    print("=" * 50)

    amount = get_valid_amount("Enter amount to withdraw: ₹")

    current_balance = accounts[account_number]["balance"]

    if amount > current_balance:
        print("\nInsufficient balance!")
        print(f"Available Balance: ₹{current_balance:.2f}")
        return

    accounts[account_number]["balance"] -= amount

    add_transaction(
        account_number,
        "WITHDRAW",
        amount,
        "Money withdrawn from account"
    )

    print("\nWithdrawal successful!")
    print(f"Withdrawn Amount : ₹{amount:.2f}")
    print(f"Remaining Balance: ₹{accounts[account_number]['balance']:.2f}")


# ============================================================
# TRANSFER MONEY
# ============================================================

def transfer_money(sender_account):
    """Transfer money from one account to another."""

    print("\n" + "=" * 50)
    print("                 MONEY TRANSFER")
    print("=" * 50)

    receiver_account = input("Enter receiver account number: ").strip()

    # Check receiver account
    if receiver_account not in accounts:
        print("Receiver account not found.")
        return

    # Prevent transfer to the same account
    if receiver_account == sender_account:
        print("You cannot transfer money to your own account.")
        return

    amount = get_valid_amount("Enter amount to transfer: ₹")

    # Check sender balance
    if amount > accounts[sender_account]["balance"]:
        print("\nInsufficient balance!")
        print(
            f"Available Balance: "
            f"₹{accounts[sender_account]['balance']:.2f}"
        )
        return

    # Deduct from sender
    accounts[sender_account]["balance"] -= amount

    # Add to receiver
    accounts[receiver_account]["balance"] += amount

    # Add sender transaction
    add_transaction(
        sender_account,
        "TRANSFER",
        amount,
        f"Transferred to account {receiver_account}"
    )

    # Add receiver transaction
    add_transaction(
        receiver_account,
        "TRANSFER RECEIVED",
        amount,
        f"Received from account {sender_account}"
    )

    print("\nTransfer successful!")
    print(f"Transferred Amount : ₹{amount:.2f}")
    print(f"Receiver Account   : {receiver_account}")
    print(
        f"Your New Balance   : "
        f"₹{accounts[sender_account]['balance']:.2f}"
    )


# ============================================================
# TRANSACTION HISTORY
# ============================================================

def transaction_history(account_number):
    """Display all transactions of an account."""

    print("\n" + "=" * 70)
    print("                    TRANSACTION HISTORY")
    print("=" * 70)

    transactions = accounts[account_number]["transactions"]

    if not transactions:
        print("No transactions found.")
        return

    for index, transaction in enumerate(transactions, start=1):

        print(f"\nTransaction {index}")
        print("-" * 50)
        print(f"Type    : {transaction['type']}")
        print(f"Amount  : ₹{transaction['amount']:.2f}")
        print(f"Date    : {transaction['date']}")
        print(f"Details : {transaction['details']}")

    print("=" * 70)


# ============================================================
# CHANGE PIN
# ============================================================

def change_pin(account_number):
    """Change the account PIN."""

    print("\n" + "=" * 50)
    print("                   CHANGE PIN")
    print("=" * 50)

    old_pin = input("Enter your current PIN: ").strip()

    # Verify old PIN
    if old_pin != accounts[account_number]["pin"]:
        print("Incorrect current PIN.")
        return

    # Enter new PIN
    while True:
        new_pin = input("Enter new 4-digit PIN: ").strip()

        if len(new_pin) != 4 or not new_pin.isdigit():
            print("PIN must contain exactly 4 digits.")
            continue

        confirm_pin = input("Confirm new PIN: ").strip()

        if new_pin != confirm_pin:
            print("PINs do not match.")
            continue

        if new_pin == old_pin:
            print("New PIN cannot be the same as the old PIN.")
            continue

        break

    accounts[account_number]["pin"] = new_pin

    add_transaction(
        account_number,
        "PIN CHANGED",
        0,
        "Account PIN changed successfully"
    )

    print("\nPIN changed successfully!")


# ============================================================
# ACCOUNT MENU
# ============================================================

def account_menu(account_number):
    """Display menu after successful login."""

    while True:

        print("\n")
        print("=" * 50)
        print("                 ACCOUNT MENU")
        print("=" * 50)
        print(f"Welcome, {accounts[account_number]['name']}!")
        print(f"Account Number: {account_number}")
        print("-" * 50)
        print("1. Check Balance")
        print("2. Deposit")
        print("3. Withdraw")
        print("4. Transfer")
        print("5. Transaction History")
        print("6. Change PIN")
        print("7. Logout")
        print("=" * 50)

        choice = input("Enter your choice (1-7): ").strip()

        if choice == "1":
            check_balance(account_number)

        elif choice == "2":
            deposit_money(account_number)

        elif choice == "3":
            withdraw_money(account_number)

        elif choice == "4":
            transfer_money(account_number)

        elif choice == "5":
            transaction_history(account_number)

        elif choice == "6":
            change_pin(account_number)

        elif choice == "7":
            print("\nLogging out...")
            print("Thank you for using our Banking System!")
            break

        else:
            print("\nInvalid choice!")
            print("Please enter a number between 1 and 7.")


# ============================================================
# MAIN MENU
# ============================================================

def main_menu():
    """Display the main banking system menu."""

    while True:

        print("\n")
        print("=" * 50)
        print("             WELCOME TO BANKING SYSTEM")
        print("=" * 50)
        print("1. Create Account")
        print("2. Login")
        print("3. Exit")
        print("=" * 50)

        choice = input("Enter your choice (1-3): ").strip()

        if choice == "1":
            create_account()

        elif choice == "2":
            login()

        elif choice == "3":
            print("\n" + "=" * 50)
            print("Thank you for using our Banking System!")
            print("Have a great day!")
            print("=" * 50)
            break

        else:
            print("\nInvalid choice!")
            print("Please enter 1, 2, or 3.")


# ============================================================
# PROGRAM START
# ============================================================

if __name__ == "__main__":
    main_menu()
```
